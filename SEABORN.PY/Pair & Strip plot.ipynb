# PAIR PLOT

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
var = sns.load_dataset('tips').head(20)
var
```

```python
sns.pairplot(var,hue='sex',markers=['*','^'])
plt.figure(figsize=(4,3))
plt.show()
```

```python
sns.pairplot(var,vars=['total_bill','tip'],hue='sex',hue_order=['Female','Male'],kind='kde',markers=['o','^'])
plt.show()
```

# STRIP PLOT

```python
var2 = sns.load_dataset('tips')
var2
```

```python
sns.stripplot(x='day',y='total_bill',data=var2,hue='sex',linewidth=1,edgecolor='red',size=5,marker="^")
plt.show()
```
