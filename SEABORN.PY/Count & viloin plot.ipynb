# count plot

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data =sns.load_dataset('tips')
data
```

```python
sns.countplot(x='sex',hue='smoker',palette= 'Accent',saturation= 0.2,data=data)
plt.show()
```

```python
sns.countplot(y='sex',hue='smoker',data=data)
plt.show()
```

# Violin Plot

```python
data
```

```python
sns.violinplot(x='day',y='total_bill',data=data)
plt.show()
```

```python
sns.violinplot(x='day',y='total_bill',data=data,hue='time',linewidth=2,palette='Dark2')
plt.show()
```

```python
sns.violinplot(x='time',y='total_bill',data=data,order=['Dinner','Lunch'],linewidth=2,palette='Dark2',inner='quart')
plt.show()
```

```python
sns.violinplot(x='day',y='total_bill',data=data,hue='sex',split=True,legend=False,)
plt.show()
```
