# BOX PLOT

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
var = sns.load_dataset("tips")
var
```

```python
sns.set(style="whitegrid")
sns.boxplot(x="day", y="total_bill",hue='sex',palette='plasma',showmeans=True,meanprops={'marker':'*','markeredgecolor':'b'},data=var)
plt.show()
```

# CAT PLOT

```python
var2 = sns.load_dataset("tips")
var2
```

```python
sns.catplot(x="day", y="total_bill",hue='sex',palette='plasma',data=var2)
plt.show()
```

# MULTIPLE PLOTS( FACET-GRID)

```python
var3 = sns.load_dataset("tips")
var3
```

```python
fg = sns.FacetGrid(var3,col='sex',hue='day')
fg.map(plt.scatter,'total_bill','tip').add_legend()      
plt.show()     
```

```python
fg = sns.FacetGrid(var3,col='day',hue='sex')
fg.map(plt.scatter,'total_bill','tip').add_legend()      
plt.show()     
```

```python
fg = sns.FacetGrid(var3,col='sex',hue='day',palette='summer')
fg.map(plt.bar,'total_bill','tip',edgecolor='red').add_legend()     
plt.show()     
```
