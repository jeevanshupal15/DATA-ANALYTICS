# Line Plot


```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

seaborn works on datasets


```python
data = sns.load_dataset('penguins')
data
 
```

```python
sns.lineplot(x='bill_length_mm',y='bill_depth_mm',data = data, hue='sex',size=50,style='sex',palette='Accent',markers=['o','>'],
             dashes=False,legend=False)
plt.title('use of seaborn')
plt.grid(True)
plt.show()
```

# Histogram

```python
sns.displot(data['bill_length_mm'],color='blue',kde=True,rug=True)
plt.show()
```

organise in order

```python
sns.displot(data['bill_length_mm'],bins=[35,40,45,50,55,60],color='blue',kde=True,rug=True)
plt.show()
```

# BAR PLOT

```python
sns.barplot(x= data.island, y=data.bill_length_mm)
plt.show()
```

Other prameters


```python
order_1 =['Biscoe','Dream','Torgersen']
sns.barplot(x= data.island, y=data.bill_length_mm,hue=data.sex,order=order_1,hue_order=['Female','Male'],palette='Accent',saturation=50
            ,err_kws={'color': 'red'},alpha=0.5)
sns.set_style('darkgrid')
plt.show()
```

Horizontal graph (it take only numerical value)
use = orient

```python
sns.barplot(x= data.bill_depth_mm, y=data.bill_length_mm,orient='h')
plt.show()
```
