# SCATTER PLOT


```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
data = sns.load_dataset('penguins')
data
 
```

```python
m = {'Female':'*','Male':'o'} 
sns.scatterplot(x='bill_length_mm', y='bill_depth_mm',hue = 'sex',style='sex',data=data,markers=m,size='sex',sizes=(100,100),legend=False)
plt.show()
```

# Heat map

Heat map work on 2D array

```python
var = np.linspace(1,10,20).reshape(5,4)
var
```

```python
sns.heatmap(var)
```

```python
data = sns.load_dataset("anagrams").head(10)
x = data.drop(columns=['attnr'])
x
```

```python
sns.heatmap(x,vmin=0,vmax=15,annot=True,cmap='YlGnBu',cbar=True)
```

```python
var1 = np.linspace(1,10,10).reshape(2,5)
var1
```

```python
sns.heatmap(var1,vmin=0,vmax=10,annot=True,cbar=False,xticklabels=False,yticklabels=False)
```

```python
var2 = np.array([['a1','a2','a3','a3','a4'],['b1','b2','b3','b4','b5']])
var2
```

```python
y ={'fontsize':20,'color':'r',}
v = sns.heatmap(var1,vmin=0,vmax=10,annot=var2,fmt='s',annot_kws=y,linewidths=5,cmap='YlGnBu',cbar=False,xticklabels=False,yticklabels=False)
v.set(xlabel='Time',ylabel='Temperature')
plt.show()
```
