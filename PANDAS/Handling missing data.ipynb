# Merge

```python
var1 =pd.DataFrame({'A':[1,2,3,4],'B':[11,22,33,44]})
var2 =pd.DataFrame({'A':[1,2,3,5],'C':[12,21,13,41]})
pd.merge(var1,var2,on='A') # it works only if the columns are the same
```

```python
pd.merge(var1,var2,on='A',how="inner") # it works only
```

```python
pd.merge(var1,var2,on='A',how='left') # it will give all the rows from var1 and var2 and fill the missing values with NaN
```

```python
pd.merge(var1,var2,on='A',how='right') # it will fill the missing values with NaN
```

```python
pd.merge(var1,var2,on='A',how='outer', indicator=True)
```

```python
var1 =pd.DataFrame({'A':[1,2,3,4],'B':[11,22,33,44]})
var2 =pd.DataFrame({'A':[1,2,3,5],'B':[12,21,13,41]})
pd.merge(var1,var2,on='A') 
```

```python
pd.merge(var1,var2,left_index=True,right_index=True,suffixes=('_name','_id'))
```

# Concate

```python
sr1 = pd.Series([1,2,3,4,5])
sr2 = pd.Series([11,22,33,44])
pd.concat([sr1,sr2])
```

```python
d1 =pd.DataFrame({'A':[1,3,4,8],'B':[11,22,33,44]})
d2 =pd.DataFrame({'A':[1,2,5,9],'C':[12,13,41,89]})
pd.concat([d1,d2],axis=1) 
```

```python
pd.concat([d1,d2],axis=0) 
```

```python
d1 =pd.DataFrame({'A':[1,3,8,5],'B':[11,22,33,44]})
d2 =pd.DataFrame({'A':[1,5],'C':[12,89]})
pd.concat([d1,d2],axis=1) 
```

```python
pd.concat([d1,d2],axis=1,join='inner') # inner = intersection ,outer = union
```

```python

pd.concat([d1,d2],axis=1,keys=('D1','D2')) # keys are used to name the columns
```

```python

pd.concat([d1,d2],axis=0,keys=('D1','D2'))  # keys are used to name the dataframes
```

```python
d1 =pd.DataFrame({'A':[1,3,8,5],}) #1D dataframe
d2 =pd.DataFrame({'A':[1,5],'C':[12,89]})#2D dataframe
pd.concat([d1,d2]) 
```

```python
d1 =pd.DataFrame({'D':[1,3,8,5],})
d2 =pd.DataFrame({'A':[1,5],'C':[12,89]})
pd.concat([d1,d2],axis=1) 
```
