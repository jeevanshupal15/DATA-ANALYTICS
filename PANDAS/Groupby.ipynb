# Groupby

```python
import pandas as pd
import numpy as np
```

```python
var = pd.DataFrame({'name': ['a','b','c','d','a','e','a','b','d','c'],'age':[5,4,7,3,8,2,9,1,0,6]})
var
```

```python
var1 = var.groupby('name')
var1
```

```python
# for show this data 
for x ,y in var1:
    print( x , y )
    print()
```

```python
# for get specific data from dataframe
var1.get_group('a')
```

```python
# for minimum and maximum ,mean
var1.min()
```

```python
var1.max()
```

```python
var1.mean()
```

```python
# convert in list
li =list(var1)
li
```

# joining two dataframes

```python
d1 =pd.DataFrame({'A':[1,3,4,8],'B':[11,22,33,44]})
d2 =pd.DataFrame({'D':[50,20],'C':[41,89]})
d1.join(d2)
```

```python
d2.join(d1)
```

```python
d2.join(d1,how='outer')
```
