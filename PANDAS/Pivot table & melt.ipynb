# Melt Function

```python
import pandas as pd
import numpy as np
```

```python
var  = pd.DataFrame({"days":[1,2,3,4,5,6],"eng": [10,60,49,35,56,34],"math":[20,50,45,32,34,56],"science":[30,40,45,32,34,56]})
var
```

```python
pd.melt(var) # vericaly reshape
```

```python
pd.melt(var,id_vars=["days"],var_name="subject",value_name="marks")
```

# Pivot 


```python
var  = pd.DataFrame({"days":[1,2,3,4,5,6],'students':["a",'b','c','d','e','f'],"eng": [10,60,49,35,56,34],"math":[20,50,45,32,34,56],"science":[30,40,45,32,34,56]})
var.pivot(index="days",columns="students")
```

```python
# get specific column data
var.pivot(index="days",columns="students",values="math")
```

# Pivot table

```python
# for operations
var.pivot_table(index="days",columns="students",aggfunc="sum")
```

```python
var  = pd.DataFrame({"days":[1,1,1,1,2,2],'students':["a",'b','a','b','a','b'],"eng": [10,60,49,35,56,34],"math":[20,50,45,32,34,56],"science":[30,40,45,32,34,56]})
var
```

```python
var.pivot_table(index="students",columns="days",aggfunc="sum",margins=True)
```
