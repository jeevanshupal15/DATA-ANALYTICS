# Airthmetic Operation

```python
import pandas as pd
```

```python
var = pd.DataFrame({"A":[1,2,3,4,5],"B":[6,7,8,9,10]})
var
```

# Addition Operation

```python
var["C"] = var["A"] + var["B"]
var
```

# Subtraction Operation


```python
var["D"] = var["A"] - var["B"]
var
```

# Multiplication Operation

```python
var["E"] = var["A"] * var["B"]
var
```

# Division Operation

```python
var['F'] = var['A'] / var['B']
var
```
