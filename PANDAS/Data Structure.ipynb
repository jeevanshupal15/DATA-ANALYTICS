## Series

```python
import pandas as pd
```

```python
x = [2,4,6,8,10,12,14,16] 

var = pd.Series(x)

print(var)
print(type(var))
print(var[5]) # get data through index

var2 =pd.Series(x,index = ['a','b','c','d','e','f','g','h'],dtype = float,name ="python") # change index 
print(var2) 

```

```python
# dictonary
dic = {"name":["python","java","c","c++","html"],"prop" : [12,13,14,15,16],"rank" :[1,2,3,4,5]}
var3 = pd.Series(dic)
var3
```

```python
s = pd.Series(12, index =[1,2,3,4,5,6,7,8])
print(s)
print(type(s))
```

```python
s1 = pd.Series(14,index = [1,2,3,4,5,6,7])
s2 = pd.Series(13,index = [1,2,3,4,5])
print(s1 + s2)
```
