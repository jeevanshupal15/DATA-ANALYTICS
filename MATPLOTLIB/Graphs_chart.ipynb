# Line chart/Line plot

```python
import matplotlib.pyplot as plt
```

# Key terms you should know

data point = x axis , y axis
x axis/y axis = x = vertical , y = horizontal
figure = like canvas
axes = graph box
plot = line graph,bar graph,scatter graph.histogram graph
marker = symbol(dot,circle,square,triangle,etc)
line style = solid,dashed,dotted
color = red,blue,green,yellow,etc
legend = label
title = title
grid    = grid
function = plot(x,y),show(),etc
parameter = values,range,etc
keyword argument = like (name = value)
dpi = resolution/clarity


```python
x = [1,2,3,4] #horizontal axsis
y = [10,20,15,25] #vertical axis
plt.plot(x,y)

plt.show()
```

```python
x = ['mon','tues','wed','thur','fri','sat','sun']
y = [10,5,15,3,20,17,25]

plt.plot(x,y)

plt.title('Backery Sales')

plt.xlabel('Days')
plt.ylabel('sales of week')

plt.show()
```

# Matplotlib functions

```python
#plt.plot(x,y , color = 'color name', linestyle = 'line_style',linewidth = 'value',marker = 'marker type',lable = 'labelname')
month = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
sales = [1000,500,1200,2000,1400,1800,2100,1500,1300,3000,2500,3400]

plt.plot(month,sales,color = 'blue',linestyle = '--',linewidth = 2,marker = 'o',label = '2025_sales')
plt.xlabel('months')
plt.ylabel('sales per month')
plt.title('monthly sales data')
plt.legend(loc = 'upper left',fontsize = 12)
plt.xticks(range(1,13),['M1','M2','M3','M4','M5','M6','M7','M8','M9','M10','M11','M12'])
plt.grid(color = 'black',linestyle = '-.',linewidth = .5)

plt.show()
```

# Bar chart

```python
#plt.bar(x,y , color = 'color name',width = 'value',lable = 'labelname')
product = ['A', 'B' , 'C,' , 'D' ]
sales = [1000,700,900,1500]

plt.bar(product,sales,color = 'orange',label='sales')
plt.xlabel('Product')
plt.ylabel('Sales')
plt.title('Sales of different products')
plt.legend()

plt.show()
```

```python
product = ['A', 'B' , 'C,' , 'D' ]
sales = [1000,700,900,1500]

plt.barh(product,sales,color = 'orange',label='sales') #horizontal bar chart
plt.xlabel('Product')
plt.ylabel('Sales')
plt.title('Sales of different products')
plt.legend()

plt.show()
```

# Pie Chart

```python
# plt.pie(value, lablel = lable_list, colur =colour_val, autopct='%1.1f%%')
region =['North','South','East','West']
revenue = [1000,600,1300,300]

plt.pie(revenue,labels=region,autopct='%1.1f%%',colors= ['skyblue','coral','lightgreen','yellow'])
plt.title('revenue contibution by region')

plt.show()
```

# Scatter Plot

```python
# plt.scatter(x,y , color = 'red', marker = 'o',label = 'label')
hours_studies = [1,2,3,4,5,6,7,8]
marks = [3,45,76,63,56,90,75,56]

plt.scatter(hours_studies,marks , color = 'red', marker = '^',label = 'student data')
plt.xlabel('hours studies')
plt.ylabel('marks')
plt.title('relationship b/w time and marks')
plt.legend(loc="upper left")
plt.grid(True)
plt.show()
```

```python
plt.scatter([1,2,3],[67,87,56] , color = 'red', marker = '^',label = 'class A') #grp 1
plt.scatter([1,2,3],[34,78,64] , color = 'blue', marker = '^',label = 'class B') #grp 2
plt.xlabel('hours studies')
plt.ylabel('marks')
plt.title('comparison of two marks')
plt.legend(loc="upper left")
plt.grid(True)
plt.show()
```

# Subplots Function
 for use multiple graphs & charts in one canvas

```python
#plt.subplot(nrows, ncols, index)

x = [1,2,3,4]
y = [10,5,12,5]

plt.subplot(1,2,1) #1row,2col,1st subplot
plt.plot(x,y)
plt.title('Line Plot')

plt.subplot(1,2,2)#1row,2col,2nd subplot
plt.bar(x,y)
plt.title('Bar Plot')

plt.show()

```

```python
#fig ,ax = plt.subplots(nrows,ncols,figsize=(width,height))
fig , ax = plt.subplots(1,2,figsize=(10,5))

x = [1,2,3,4]
y = [10,5,12,5]

ax[0].plot(x,y)
ax[0].set_title('Line Plot')

ax[1].bar(x,y)
ax[1].set_title('Bar Plot')

plt.suptitle('comparison of line and bar plot')
plt.tight_layout() #to make the plot fit in the window
plt.show()
```
