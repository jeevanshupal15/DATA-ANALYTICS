# Advanced operration with Business Data

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data structures : [resturant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([[1, 100, 200, 300, 400],   # Resturant 1
                       [2, 200, 400, 600, 800],   # Resturant 2
                       [3, 300, 600, 900, 1200],  # Resturant 3
                       [4, 400, 800, 1200, 1600], # Resturant 4
                       [5, 500, 1000, 1500, 2000]]) # Resturant 5

print("========== zomato_sales_analysis ==========")
print("\n sales_data shape :", sales_data.shape )
print(" first 3 resturant sales",sales_data[0:3])
```

```python
# total sales per year
yearly_total = np.sum(sales_data[:,1:], axis=0)
print(yearly_total)

# min sales per resturant 
min_sales = np.min(sales_data[:,1:],axis=1)
print(min_sales)

# max sales per yearly
max_sales = np.max(sales_data[:,1:],axis=0)
print(max_sales)

#average sales per resturant
avg_sales =np.mean(sales_data[:,1:],axis=1)
print(avg_sales)

#cumlative sales per resturant
cum_sales = np.cumsum(sales_data[:,1:],axis=1)
print(cum_sales)
```

# Graph of average sales

```python
# graph of average
plt.figure(figsize=(10,6))
plt.plot(np.mean(cum_sales,axis=0))
plt.title("average sales across all resturants")
plt.xlabel("years")
plt.ylabel("sales")
plt.grid(True)
plt.show()
```

```python
vec1 = np.array([1,2,3,4,5])
vec2 = np.array([6,7,8,9,10])

print("vector addition",vec1 + vec2)
print("vector subtraction",vec1 - vec2)
print("vector multiplication",vec1 * vec2)
print("vector division",vec1 / vec2)

print("vector dot product",np.dot(vec1,vec2))
angle = np.arccos(np.dot(vec1,vec2)/(np.linalg.norm(vec1)*np.linalg.norm(vec2))) 
print("angle between the vectors",angle)

```
